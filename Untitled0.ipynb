```python
d3['saison_club']=0
```

```python
for n in d3['club_name']:
    for m in d2['common_name']:
        for l in d2['season'].unique():
            if m == n:
                a = str(d2['league_position'][(d2['common_name']==m)&(d2['season']==l)])
                if (a[1:2] == ' '):
                    if (a[6:7] == '\\'):
                        d3.loc[(d3['club_name']==m)&(d3['season']==l),'saison_club'] = int(a[5:6])
                    else:
                        d3.loc[(d3['club_name']==m)&(d3['season']==l),'saison_club'] = int(a[5:7])
                else:
                    if (a[1:2] == 'e'):
                        continue
                    if (a[2:3] != ' '):
                        if (a[8:9] == '\\'):
                            d3.loc[(d3['club_name']==m)&(d3['season']==l),'saison_club'] = int(a[7:8])
                            continue
                        else:
                            d3.loc[(d3['club_name']==m)&(d3['season']==l),'saison_club'] = int(a[7:9])
                            continue
                    if (a[7:8] == '\\'):
                        d3.loc[(d3['club_name']==m)&(d3['season']==l),'saison_club'] = int(a[6:7])
                    else:
                        d3.loc[(d3['club_name']==m)&(d3['season']==l),'saison_club'] = int(a[6:8])
```

```python
d3['saison_club_inolved']=0
```

```python
for n in d3['club_involved_name']:
    for m in d2['common_name']:
        for l in d2['season'].unique():
            if m == n:
                a = str(d2['league_position'][(d2['common_name']==m)&(d2['season']==l)])
                if (a[1:2] == ' '):
                    if (a[6:7] == '\\'):
                        d3.loc[(d3['club_involved_name']==m)&(d3['season']==l),'saison_club_involved'] = int(a[5:6])
                    else:
                        d3.loc[(d3['club_involved_name']==m)&(d3['season']==l),'saison_club_involved'] = int(a[5:7])
                else:
                    if (a[1:2] == 'e'):
                        continue
                    if (a[2:3] != ' '):
                        if (a[8:9] == '\\'):
                            d3.loc[(d3['club_involved_name']==m)&(d3['season']==l),'saison_club_involved'] = int(a[7:8])
                            continue
                        else:
                            d3.loc[(d3['club_involved_name']==m)&(d3['season']==l),'saison_club_involved'] = int(a[7:9])
                            continue
                    if (a[7:8] == '\\'):
                        d3.loc[(d3['club_involved_name']==m)&(d3['season']==l),'saison_club_involved'] = int(a[6:7])
                    else:
                        d3.loc[(d3['club_involved_name']==m)&(d3['season']==l),'saison_club_involved'] = int(a[6:8])
```
